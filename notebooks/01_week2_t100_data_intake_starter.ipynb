Goal: discover the downloaded annual ZIPs, inspect one archive, load a small sample, and verify the available fields without modifying the raw files.


```python
from pathlib import Path
import sys

# Make src/ importable when the notebook is run from the repository root.
repo_root = Path.cwd()
if repo_root.name == 'notebooks':
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root / 'src'))

from t100_io import list_t100_zips, archive_members, read_t100_zip, normalize_columns, core_field_check
```

```python
zips = list_t100_zips(repo_root / 'data' / 'raw')
print(f'Annual ZIPs found: {len(zips)}')
for p in zips:
    print(p.name)
```

```python
if not zips:
    raise FileNotFoundError('No T-100 ZIPs found. Download the raw data first.')

sample_zip = zips[-1]
print('Inspecting:', sample_zip.name)
archive_members(sample_zip)[:10]
```

```python
sample = normalize_columns(read_t100_zip(sample_zip, nrows=5000))
print('Sample shape:', sample.shape)
display(sample.head())
```

```python
check = core_field_check(sample)
print('Core fields found:', check['found'])
print('Core fields missing:', check['missing'])
print('
All columns:')
print(sample.columns.tolist())
```

```python
summary = sample.dtypes.astype(str).rename('dtype').to_frame()
summary['missing_n'] = sample.isna().sum()
summary['missing_pct'] = (sample.isna().mean() * 100).round(2)
display(summary)
```
